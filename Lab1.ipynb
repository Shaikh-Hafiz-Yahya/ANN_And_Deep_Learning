# Lab3 Task1: Single Layer Perceptron

```python
# import required library
import numpy as np
```

```python
x = np.arange(1,4)
w = np.array([0.4 , 0.5 , 0.6])
b = 0.1
# weighted sum
z = np.dot(w , x) + b
```

### sigmoid Activation function

```python
def sigmoid(z):
    return 1 / (1+np.exp(-z))
y = sigmoid(z)
print("Weighted sum(z)=" , z)
print("Neuron Output(y)=" , y)
```
